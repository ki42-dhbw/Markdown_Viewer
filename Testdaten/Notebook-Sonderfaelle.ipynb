````python
from dataclasses import dataclass

@dataclass
class Messpunkt:
    name: str
    wert: float

def beschreibe(p: Messpunkt, einheit: str = "mm") -> str:
    # Ausgabe mit einer Nachkommastelle
    return f"{p.name}: {p.wert:.1f} {einheit}"

print(beschreibe(Messpunkt("Probe A", 12.345)))
````
